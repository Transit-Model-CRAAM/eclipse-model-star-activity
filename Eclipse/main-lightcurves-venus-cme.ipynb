# Comparando curvas de luz (com CME e Sem CMEs)
## Proposta do Script
Simular de forma direta a curva de luz de uma estrela com um evento de ejeção de massa, usando como base para matriz estrela, imagens do Sol de arquivos `.fits` do dia em que foi capturado o trânsito de Vênus em frente ao Sol.
Utilizamos a comparação do momento com evento e sem evento, para melhor analisar o Sinal da curva de luz

_Utilizamos a modelagem do planeta com os mesmos dados de vênus._

### Imports

```python
import numpy as np
from matplotlib import pyplot
from Star.Estrela import Estrela
from Planet.Eclipse import Eclipse
from Planet.Planeta import Planeta
import numpy as np

%matplotlib tk
```

# Dados estrela

```python
raio_estrela_pixel = 373. # default (pixel)
intensidade_maxima = 240 # default
tamanho_matriz = 856 # default
raio_estrela = 1 # raio da estrela em relacao ao raio do sol
coeficiente_um = 0.29
coeficiente_dois = 0.34
```

# Dados planeta
### VENUS — parâmetros orbitais e físicos


```python
periodo = 224.701 # em dias
# Atenção: escolha a inclinação conforme seu modelo:

angulo_inclinacao = 89.8 # para testes
#angulo_inclinacao = 89.95 

ecc = 0.006772             # excentricidade
anomalia = 0.0             # pode manter 0 se não estiver usando época específica

raio_plan_Jup = 0.3  # para testes
#raio_plan_Jup = 0.08656    # R_venus / R_Jup ≈ 6051.8 km / 69911 km

semi_eixo_UA = 0.723332    # UA
mass_planeta = 0.002564    # M_venus / M_Jup ≈ 4.867e24 / 1.898e27

```

# 💥 Modelando a Estrela com CME 

```python
estrela_ = Estrela(raio_estrela_pixel, raio_estrela, intensidade_maxima, coeficiente_um, coeficiente_dois, tamanho_matriz, useFits = True, fits_path="2012-06-05-venus")
tamanho_matriz = estrela_.getTamanhoMatriz()

Nx = estrela_.getNx() #Nx e Ny necessarios para a plotagem do eclipse
Ny = estrela_.getNy()
dtor = np.pi/180.  
```

## Modelando o Planeta

```python
planeta_ = Planeta(semi_eixo_UA, raio_plan_Jup, periodo, angulo_inclinacao, ecc, anomalia, estrela_.getRaioSun(), mass_planeta)

print(planeta_.getRaioPlan())
```

```python
estrela_matriz = estrela_.getMatrizEstrela()
estrela_.Plotar(tamanho_matriz, estrela_matriz)
```

## Eclipse com CME


```python
eclipse_ = Eclipse(Nx, Ny, raio_estrela_pixel, estrela_, planeta_, 1)
estrela_.Plotar(tamanho_matriz, estrela_matriz)

tempoHoras = 1
eclipse_.geraTempoHoras(tempoHoras)
eclipse_.criarEclipse(anim=True)

print ("Tempo Total (Trânsito):", eclipse_.getTempoTransito()) 
tempoTransito = eclipse_.getTempoTransito()
curvaLuz = eclipse_.getCurvaLuz()
tempoHoras = eclipse_.getTempoHoras()
curvaLuz_1 = curvaLuz

#Plotagem da curva de luz 
pyplot.plot(tempoHoras, curvaLuz)
pyplot.axis([-tempoTransito/2, tempoTransito/2, min(curvaLuz)-0.001, 1.001])                       
pyplot.show()
```

```python
latsugerida = eclipse_.calculaLatMancha()
```

# ☀ Modelando Estrela Sem CME

```python
estrela_sem_cme = Estrela(raio_estrela_pixel, raio_estrela, intensidade_maxima, coeficiente_um, coeficiente_dois, tamanho_matriz, useFits = True, fits_path="2012-06-05-venus-no-cme")
tamanho_matriz = estrela_sem_cme.getTamanhoMatriz()

Nx = estrela_sem_cme.getNx() 
Ny = estrela_sem_cme.getNy()
dtor = np.pi/180.  
```

```python
planeta_ = Planeta(semi_eixo_UA, raio_plan_Jup, periodo, angulo_inclinacao, ecc, anomalia, estrela_sem_cme.getRaioSun(), mass_planeta)

print(planeta_.getRaioPlan())
```

```python
estrela_matriz = estrela_sem_cme.getMatrizEstrela()
estrela_sem_cme.Plotar(tamanho_matriz, estrela_matriz)
```

```python
eclipse_ = Eclipse(Nx, Ny, raio_estrela_pixel, estrela_sem_cme, planeta_, 2)
estrela_sem_cme.Plotar(tamanho_matriz, estrela_matriz)

tempoHoras = 1
eclipse_.geraTempoHoras(tempoHoras)
eclipse_.criarEclipse(anim=True)

print ("Tempo Total (Trânsito):", eclipse_.getTempoTransito()) 
tempoTransito = eclipse_.getTempoTransito()
curvaLuz = eclipse_.getCurvaLuz()
tempoHoras = eclipse_.getTempoHoras()
curvaLuz_2 = curvaLuz
```

# 📈 Comparando resultados

### Plotagem da curva de luz 

```python
pyplot.plot(tempoHoras, curvaLuz_1, color="red", label="Curva de Luz com CME")
pyplot.plot(tempoHoras, curvaLuz_2, color="black", label="Curva de Luz sem CME")

pyplot.xlabel("Tempo (horas)")
pyplot.ylabel("Intensidade Normalizada")
pyplot.legend()  # Mostra a legenda
pyplot.title("Comparação entre curvas de luz com e sem CME do dia 2012/06/05 (Trânsito de Vênus)")
pyplot.grid(True)
pyplot.show()

pyplot.axis([-tempoTransito/2, tempoTransito/2, min(curvaLuz_1)-0.001, 1.001])                       
pyplot.show()
```

# Ruído

```python
ruido = np.array(curvaLuz_1) - np.array(curvaLuz_2)

pyplot.plot(tempoHoras, ruido, label='Ruído da CME', alpha=0.5)
pyplot.xlabel('Tempo')
pyplot.xticks(rotation=65)
pyplot.ylabel('Fluxo Normalizado')
pyplot.legend()
pyplot.show()
```
